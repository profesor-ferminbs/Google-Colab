### **Carga de datos -> Función open()**
Carga de datasets muy grandes. La idea es no leer todo de golpe, sino de forma gradual. Borra los datos cargados en memoria una vez se han procesado.

```python
import urllib.request

# La URL de internet usa %20 para representar los espacios
url = "https://raw.githubusercontent.com/profesor-ferminbs/Google-Colab/Curso_Data_Science/Customer%20Churn%20Model.txt"
# Guardamos el archivo localmente con espacios en el nombre
urllib.request.urlretrieve(url, "Customer Churn Model.txt")

# Abrir con la función open() pasando el nombre exacto con espacios. Modo lectura "r"
data2 = open("Customer Churn Model.txt", "r")
```

```python
#Leer la siguiente línea. Quitamos espacios al inicio y final y separamos por comas.
cols = data2.readline().strip().split(",")
cols
```

```python
n_cols = len(cols)
n_cols
```

```python
counter = 0
main_dict ={}
for col in cols:
    main_dict[col] = []
main_dict
```

```python
for line in data2:
    values = line.strip().split(",")
    for i in range(len(cols)):
        main_dict[cols[i]].append(values[i])
    counter += 1
print ("%d Filas y %d Columnas"%(counter, n_cols))
```

```python
df2 = pd.DataFrame(main_dict)
df2.head()
#Mismo resultado que read_csv pero lo hemos hecho de forma manual para probar otro sistema y cargar de forma gradual.
```
